###Data Ingesion

```python
import os 
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_texts ={
    "../data/text_files/python_intro.txt": """Python is a popular, high-level,
    and interpreted programming language known for its clean syntax, readability, 
    and versatility.Key Features of PythonEasy to Learn: Uses simple, English-like words
    and indentation instead of brackets.Interpreted: Executes code line by line,
    making it easy to test and debug.Cross-Platform: Runs on Windows, Mac,
    and Linux without changing code.Large Libraries: Offers powerful built-in tools
    and external packages for diverse tasks.Multi-Purpose: Used widely in web development,
    data science, AI, and automation.""",

    "../data/text_files/ml_intro.txt": """Machine learning (ML) is a core branch of artificial 
    intelligence focused on building systems that learn from data, identify patterns, 
    and make decisions without explicit programming.
    Key Aspects of Machine LearningDefinition: A subset of AI where computers use algorithms and 
    statistical models to analyze data.Data-Driven: Systems improve automatically through
     experience and by processing more input data.Core 
     Types: Divided into supervised learning (labeled data),
      unsupervised learning (hidden patterns), and reinforcement learning (trial and error).
      Real-World Use: Powers everyday tools like email spam filters, recommendation engines,
       and voice assistants.If you want, I can dive deeper into supervised vs. unsupervised learning
        or list some popular ML algorithms like linear regression and decision trees."""
}

for filepath, content in sample_texts.items():
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(content)
print("Sample text files created successfully.")
```

```python
###TextLoader
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")

documents = loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader("../data/text_files",
                              glob="**/*.txt",
                              loader_cls=TextLoader, 
                              loader_kwargs={"encoding": "utf-8"},
                              show_progress=False)

documents = dir_loader.load()
documents
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyMuPDFLoader

dir_loader = DirectoryLoader(
    "../data/pdf_files",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=False
)

pdf_documents = dir_loader.load()
pdf_documents
```
